# Panel Judge — FULL DATASET RUN v6 (Patched + H1/H5 Pilot Iteration 2)

Three-model panel (Claude Opus 4.7, GPT-5.5, Gemini 3.1 Pro) cross-judging hallucinations
on UCF-Crime model outputs.

## What's in this notebook

**Cells 2–24:** Original v4 main panel. Untouched.
**Cells 26–34 (v5):** First pilot — disambiguated H1/H5 prompt. Result: GPT and Gemini
both flagged ~100% of rows positive on H1/H5 (no variance, kappa undefined). Also,
the first 200 rows happened to all be Claude-generated, so Claude never judged.
**Cells 36–44 (v6 NEW):** Second pilot — middle-ground H1/H5 definitions plus a
**stratified sample** that includes rows from all three generator models, so all
three judges contribute verdicts.

## v6 design changes from v5

1. **Middle-ground prompt** (`JUDGE_SYSTEM_V3`): H1/H5 defined as
   "fabrication that adds a *meaningfully new* claim beyond ground truth, distinct
   from descriptive synonyms or generic specifications." With paired examples for
   each direction.
2. **Stratified pilot sample**: sample ~67 rows from each generator model with
   `random_state=42` for reproducibility. Saves to a new subfolder.
3. **Pilot output**: `OUTPUT_DIR / 'h1_h5_v3' /` (separate from v5's folder).

## Decision criteria for v6 pilot

After running cells 36–44 once, look at:
- **Positive rates per judge.** Each judge should be in 10-70% range on H1/H5
  (not 0% or 100%). If still pinned at extremes, the prompt is still mis-calibrated.
- **Kappa across all three pairs.** Need all three pairs (Claude-GPT, Claude-Gemini,
  GPT-Gemini) at >= 0.4 to consider committing to a full re-run.

## 2. Auto-clean retryable errors (preserve permanent ones)

````python
# Auto-cleanup of bad checkpoint entries.
# Permanent policy blocks (Gemini PROHIBITED_CONTENT) are preserved — retrying them
# is futile, and we want the run to terminate. Other -1 entries are removed so they
# get retried with the patched judge functions.

if os.path.exists(PANEL_CHECKPOINT):
    with open(PANEL_CHECKPOINT) as f:
        _ckpt = json.load(f)
    _initial = len(_ckpt)

    _bad_keys = []
    _kept_permanent = 0
    for k, v in _ckpt.items():
        if not isinstance(v, dict):
            continue
        has_neg1 = any(v.get(h, 0) == -1 for h in ['H1','H2','H3','H4','H5','H6'])
        if not has_neg1:
            continue
        reasoning = str(v.get('reasoning', ''))
        # Don't retry hard policy blocks — they will never succeed
        if 'PROHIBITED_CONTENT' in reasoning:
            _kept_permanent += 1
            continue
        _bad_keys.append(k)

    for k in _bad_keys:
        del _ckpt[k]

    if _bad_keys:
        with open(PANEL_CHECKPOINT, 'w') as f:
            json.dump(_ckpt, f)
        print(f'Auto-cleanup: removed {len(_bad_keys)} retryable errors')
        print(f'  Checkpoint: {_initial} -> {len(_ckpt)} entries')
    else:
        print(f'Auto-cleanup: nothing retryable to clean ({_initial} entries)')
    if _kept_permanent:
        print(f'  Preserved {_kept_permanent} permanent policy-block entries (will not retry)')
else:
    print('No existing full checkpoint - starting fresh.')
````

## 3. Ground truth

````python
# Ground truth loader
GT_FILES = ['UCFCrime_Train.json', 'UCFCrime_Val.json', 'UCFCrime_Test.json']

def derive_crime_type(video_id):
    name = video_id.replace('_x264', '')
    m = re.match(r'^([A-Za-z]+?)\d', name)
    return m.group(1) if m else 'Unknown'

def is_anomalous(video_id):
    return not video_id.startswith('Normal_')

def parse_ground_truth(gt_base_path):
    gt_data = {}
    for fname in GT_FILES:
        fpath = os.path.join(gt_base_path, fname)
        if not os.path.exists(fpath):
            continue
        with open(fpath, 'r', encoding='utf-8') as f:
            data = json.load(f)
        for video_id, info in data.items():
            if not is_anomalous(video_id):
                continue
            gt_data[video_id] = {
                'crime_type': derive_crime_type(video_id),
                'sentences':  info.get('sentences', []),
                'timestamps': info.get('timestamps', []),
                'duration':   info.get('duration'),
            }
    return gt_data

if os.path.exists(GT_CACHE_PATH):
    with open(GT_CACHE_PATH) as f:
        gt_data = json.load(f)
    print(f'Loaded GT from cache: {len(gt_data):,} videos')
else:
    gt_data = parse_ground_truth(GT_PATH)
    with open(GT_CACHE_PATH, 'w') as f:
        json.dump(gt_data, f)
    print(f'Cached {len(gt_data):,} GT videos')

# Video ID helpers
GEMINI_VID_RE = re.compile(r'^([A-Za-z]+_[A-Za-z]+\d+(?:_x264)?)')

def video_id_from_filename(filename):
    m = GEMINI_VID_RE.match(filename)
    return m.group(1) if m else None

def resolve_gt_key(video_id, gt_data):
    candidates = [video_id, video_id.replace('_x264', '')]
    parts = video_id.split('_')
    if len(parts) >= 2:
        wp = '_'.join(parts[1:])
        candidates += [wp, wp.replace('_x264', '')]
    for c in candidates:
        if c in gt_data:
            return c
    return None

# Gemini extractors (only used by Stage 1 to refresh new Gemini videos)
def extract_gemini_sequential_obj(obj):
    results = obj.get('sequential_results', obj)
    if not isinstance(results, dict): return ''
    final = results.get('Final Synthesis', {})
    if isinstance(final, dict):
        text = final.get('response', '')
        if text and text.strip(): return text.strip()
    if isinstance(final, str) and final.strip(): return final.strip()
    step_keys = sorted(
        [k for k in results if k.startswith('Step')],
        key=lambda x: int(x.split()[-1]) if x.split()[-1].isdigit() else 0,
        reverse=True
    )
    if step_keys:
        val = results[step_keys[0]]
        if isinstance(val, dict): return val.get('response', '')
        return str(val)
    return ''

def extract_gemini_ltm_obj(obj):
    results = obj.get('least_to_most_results', obj)
    if not isinstance(results, dict): return ''
    step_keys = sorted(
        [k for k in results if k.startswith('Step')],
        key=lambda x: int(x.split()[-1]) if x.split()[-1].isdigit() else 0,
        reverse=True
    )
    if not step_keys: return ''
    last = results[step_keys[0]]
    if isinstance(last, dict): return last.get('response', '')
    return last if isinstance(last, str) else ''

def load_gemini_running_tech(tech_folder):
    """Read all currently-completed Gemini videos for one in-progress technique."""
    tech_path = os.path.join(BASE_PATHS['Gemini'], tech_folder)
    if not os.path.isdir(tech_path):
        return {}
    extractor = (extract_gemini_sequential_obj if tech_folder == 'SEQUENTIAL'
                 else extract_gemini_ltm_obj)
    all_files = sorted([f for f in os.listdir(tech_path)
                        if f.endswith('.json') and '_checkpoints' not in f])
    complete  = [f for f in all_files if '_complete_' in f]
    per_video = defaultdict(list)
    for fname in complete:
        try:
            with open(os.path.join(tech_path, fname), encoding='utf-8') as f:
                obj = json.load(f)
        except Exception:
            continue
        text = extractor(obj)
        vid = video_id_from_filename(fname)
        if vid and text and text.strip():
            per_video[vid].append(text.strip())
    return {vid: texts[-1] for vid, texts in per_video.items()}

print('Helpers ready.')
````

## 4. Build / refresh triplets

````python
if not os.path.exists(TRIPLETS_CACHE_PATH):
    raise FileNotFoundError(
        f'{TRIPLETS_CACHE_PATH} missing. Run the original build_triplets cell first.'
    )

df_old = pd.read_csv(TRIPLETS_CACHE_PATH)
print(f'Existing cache: {len(df_old):,} rows')
print('  Per (model, technique):')
print(df_old.groupby(['model', 'technique']).size().unstack(fill_value=0))

existing_keys = set(zip(df_old['model'], df_old['technique'], df_old['video']))

RUNNING_TECHS = {
    'SEQUENTIAL':    'Sequential',
    'LEAST-TO-MOST': 'Least-to-Most',
}

new_rows = []
for tech_folder, tech_label in RUNNING_TECHS.items():
    outputs = load_gemini_running_tech(tech_folder)
    on_disk = len(outputs)
    added = 0
    for video_id, output_text in outputs.items():
        gt_key = resolve_gt_key(video_id, gt_data)
        if gt_key is None:
            continue
        key = ('Gemini', tech_label, gt_key)
        if key in existing_keys:
            continue
        gt_info = gt_data[gt_key]
        new_rows.append({
            'model':                 'Gemini',
            'technique':             tech_label,
            'video':                 gt_key,
            'crime_type':            gt_info['crime_type'],
            'ground_truth':          ' '.join(gt_info['sentences']),
            'model_output':          output_text,
            'model_output_full_len': len(output_text),
        })
        added += 1
    print(f'  [Gemini/{tech_label}] on disk: {on_disk}  NEW added: {added}')

if new_rows:
    df_new = pd.DataFrame(new_rows)
    df_triplets = pd.concat([df_old, df_new], ignore_index=True)
    df_triplets.to_csv(TRIPLETS_CACHE_PATH, index=False)
    print(f'\nCache: {len(df_old):,} -> {len(df_triplets):,}  (+{len(new_rows)} new)')
else:
    df_triplets = df_old
    print('\nNo new Gemini videos found — cache already up to date.')

print('\nFinal triplet pool per (model, technique):')
print(df_triplets.groupby(['model', 'technique']).size().unstack(fill_value=0))
print(f'\nTotal rows: {len(df_triplets):,}')
print(f'Total judge calls if fresh: {len(df_triplets) * 2:,}')
````

## 5. SDK clients + prompt + retry helpers

````python
import anthropic
import openai
from google import genai as google_genai
from google.genai import types as google_genai_types

_anthropic_client = anthropic.Anthropic(api_key=ANTHROPIC_API_KEY)
_openai_client    = openai.OpenAI(api_key=OPENAI_API_KEY)
# Gemini Developer API (not Vertex AI) - plain API key
_genai_client     = google_genai.Client(
    api_key=os.environ['GOOGLE_API_KEY'],
)

JUDGE_SYSTEM = """You are an expert forensic video analysis evaluator.
Your task is to detect hallucinations in an AI model's analysis of a crime video.
You will be given:
  1. GROUND TRUTH: the human-annotated description of what actually happened.
  2. MODEL OUTPUT: the AI system's analysis of the same video.

Evaluate whether the MODEL OUTPUT contains any of the following hallucination types.
For each type respond ONLY with 0 (not present) or 1 (present).

Hallucination types:
  H1 SCENE_FABRICATION    - invents setting details not in ground truth (wrong location, objects, environment)
  H2 CRIME_MISCLASSIFICATION - identifies the wrong crime type (e.g. says Robbery when it is Assault)
  H3 CRIME_MISSED         - fails to mention the primary crime that is clearly described in ground truth
  H4 SEVERITY_MINIMIZATION - describes crime as less serious than ground truth indicates
  H5 ENTITY_FABRICATION   - invents people, vehicles, or objects not present in ground truth
  H6 PHANTOM_ACTORS       - adds perpetrators or victims not mentioned in ground truth

Respond ONLY with a JSON object in this exact format, no extra text:
{
  "H1": 0,
  "H2": 0,
  "H3": 0,
  "H4": 0,
  "H5": 0,
  "H6": 0,
  "reasoning": "brief one-sentence explanation"
}"""

def make_judge_prompt(ground_truth: str, model_output: str) -> str:
    gt_trunc  = ground_truth[:1500]
    out_trunc = model_output[:3000]
    return (
        f"GROUND TRUTH:\n{gt_trunc}\n\n"
        f"MODEL OUTPUT:\n{out_trunc}\n\n"
        "Now evaluate for hallucinations and return the JSON."
    )

def _retry_call(fn, max_retries=7, base_delay=4.0, max_delay=300.0):
    last_err = None
    for attempt in range(max_retries):
        try:
            return fn()
        except Exception as e:
            last_err = e
            err_str  = str(e)
            err_low  = err_str.lower()
            is_rate    = any(x in err_low for x in ['429', 'rate limit', 'quota', 'resource exhausted', 'resource_exhausted'])
            is_server  = any(x in err_low for x in ['500', '502', '503', '504', 'server error', 'unavailable', 'overloaded'])
            is_timeout = any(x in err_low for x in ['timeout', 'timed out', 'deadline'])
            is_truncated = 'truncated response' in err_low
            retryable = is_rate or is_server or is_timeout or is_truncated
            if retryable and attempt < max_retries - 1:
                delay = min(base_delay * (2 ** attempt), max_delay)
                short_err = err_str[:250].replace('\n', ' ')
                tqdm.write(f'    [{type(e).__name__} attempt {attempt+1}/{max_retries}] sleeping {delay:.0f}s | {short_err}')
                time.sleep(delay)
            else:
                raise
    if last_err:
        raise last_err
    return None

def parse_judge_response(text: str) -> dict:
    """Parse a judge JSON response, with regex fallback for malformed output.

    Patched: catches json.JSONDecodeError specifically (not all ValueErrors)
    so the regex fallback actually runs for malformed JSON like single-quoted
    keys, trailing commas, or unescaped quotes in the reasoning field.
    """
    text = re.sub(r'```(?:json)?', '', text).strip().rstrip('`').strip()

    # Try strict JSON first
    obj = None
    try:
        obj = json.loads(text)
    except json.JSONDecodeError:
        pass  # fall through to regex fallback

    if obj is not None:
        labels = {}
        for h in ['H1','H2','H3','H4','H5','H6']:
            val = obj.get(h, -1)
            labels[h] = int(bool(val)) if val in (0, 1, True, False) else -1
        labels['reasoning'] = str(obj.get('reasoning', ''))
        n_missing = sum(1 for h in ['H1','H2','H3','H4','H5','H6'] if labels[h] == -1)
        if n_missing > 0:
            raise ValueError(f'Truncated response: {n_missing} of 6 fields missing')
        return labels

    # Regex fallback for malformed JSON (single quotes, trailing commas, etc.)
    labels = {}
    for h in ['H1','H2','H3','H4','H5','H6']:
        m = re.search(r'["\']?' + h + r'["\']?\s*:\s*([01])', text)
        labels[h] = int(m.group(1)) if m else -1
    rm = re.search(r'["\']?reasoning["\']?\s*:\s*["\']([^"\']+)["\']', text)
    labels['reasoning'] = rm.group(1) if rm else 'parsed via regex fallback'
    n_missing = sum(1 for h in ['H1','H2','H3','H4','H5','H6'] if labels[h] == -1)
    if n_missing > 0:
        raise ValueError(f'Truncated response: {n_missing} of 6 fields missing')
    return labels

print('Prompt + retry helper + patched parse_judge_response ready.')
````

## 6. Judge functions (patched)

````python
def _extract_text_from_anthropic_response(resp):
    """Find the text block in an Anthropic response, skipping any thinking blocks.

    On Claude Opus 4.7, the response stream can include an empty thinking block
    as content[0] even when adaptive thinking is off (display defaults to "omitted").
    The actual answer text is in a later block with type == "text".
    """
    for block in resp.content:
        if getattr(block, 'type', None) == 'text':
            txt = getattr(block, 'text', None)
            if txt:
                return txt
    parts = []
    for block in resp.content:
        t = getattr(block, 'text', None)
        if t:
            parts.append(t)
    if parts:
        return '\n'.join(parts)
    raise ValueError(
        f'No text block in Anthropic response. '
        f'Blocks: {[getattr(b, "type", "?") for b in resp.content]}'
    )


def call_claude_judge(ground_truth: str, model_output: str) -> dict:
    prompt = make_judge_prompt(ground_truth, model_output)
    def _call():
        resp = _anthropic_client.messages.create(
            model='claude-opus-4-7',
            max_tokens=8192,
            system=JUDGE_SYSTEM,
            messages=[{'role': 'user', 'content': prompt}],
            # NOTE: no temperature - Opus 4.7 returns 400 on non-default temp/top_p/top_k
        )
        return _extract_text_from_anthropic_response(resp)
    raw = _retry_call(_call)
    return parse_judge_response(raw)


def call_gpt_judge(ground_truth: str, model_output: str) -> dict:
    prompt = make_judge_prompt(ground_truth, model_output)
    def _call():
        resp = _openai_client.chat.completions.create(
            model='gpt-5.5',
            messages=[
                {'role': 'system', 'content': JUDGE_SYSTEM},
                {'role': 'user',   'content': prompt},
            ],
            max_completion_tokens=8192,
        )
        return resp.choices[0].message.content
    raw = _retry_call(_call)
    return parse_judge_response(raw)


def call_gemini_judge(ground_truth: str, model_output: str) -> dict:
    prompt = make_judge_prompt(ground_truth, model_output)
    full_prompt = JUDGE_SYSTEM + '\n\n' + prompt

    # Disable output safety filters - UCF-Crime descriptions trip BLOCK_MEDIUM and
    # cause resp.text to return None. We're evaluating, not generating new content.
    # Note: the input PROHIBITED_CONTENT filter is a hard policy block that cannot
    # be disabled at any threshold; those failures are permanent and surfaced below.
    _safety = [
        google_genai_types.SafetySetting(category=c, threshold='BLOCK_NONE')
        for c in (
            'HARM_CATEGORY_HARASSMENT',
            'HARM_CATEGORY_HATE_SPEECH',
            'HARM_CATEGORY_SEXUALLY_EXPLICIT',
            'HARM_CATEGORY_DANGEROUS_CONTENT',
        )
    ]
    _gemini_config = google_genai_types.GenerateContentConfig(
        temperature=0.1,
        # Gemini 3.1 Pro requires thinking mode; we can't disable it. Instead, we give
        # enough total output budget for thinking + JSON to coexist. Diagnostics showed
        # thinking consumed ~975 tokens on Fighting prompts, leaving the 1024 budget
        # cut off mid-JSON. 4096 leaves comfortable room for both.
        max_output_tokens=8192,
        top_p=0.8,
        top_k=10,
        safety_settings=_safety,
    )

    def _call():
        try:
            resp = _genai_client.models.generate_content(
                model='gemini-3.1-pro-preview',
                contents=full_prompt,
                config=_gemini_config,
            )
        except Exception as e:
            err = str(e)
            if '429' in err or 'RESOURCE_EXHAUSTED' in err or 'quota' in err.lower():
                tqdm.write('    [Gemini 429] sleeping 60s before retry...')
                time.sleep(60)
            raise

        # resp.text is None when blocked / no candidates / no parts.
        # Surface a clear reason instead of letting parse_judge_response choke on None.
        text = resp.text
        if text is None:
            finish = None
            block_reason = None
            try:
                if resp.candidates:
                    finish = getattr(resp.candidates[0], 'finish_reason', None)
                if getattr(resp, 'prompt_feedback', None):
                    block_reason = getattr(resp.prompt_feedback, 'block_reason', None)
            except Exception:
                pass
            raise ValueError(
                f'Gemini returned no text (finish_reason={finish}, block_reason={block_reason})'
            )
        return text

    raw = _retry_call(_call)
    return parse_judge_response(raw)


JUDGE_CALL_MAP = {
    'Claude': call_claude_judge,
    'GPT':    call_gpt_judge,
    'Gemini': call_gemini_judge,
}

print('Judge functions ready.')
````

````python
def load_checkpoint():
    if os.path.exists(PANEL_CHECKPOINT):
        with open(PANEL_CHECKPOINT) as f:
            return json.load(f)
    return {}

def save_checkpoint(ckpt):
    with open(PANEL_CHECKPOINT, 'w') as f:
        json.dump(ckpt, f)

def run_judge_panel(df, max_workers=8):
    from concurrent.futures import ThreadPoolExecutor, as_completed
    import threading

    ckpt = load_checkpoint()
    ckpt_lock = threading.Lock()
    rows_lock = threading.Lock()
    rows_out = []
    tasks = []
    cached_rows = []
    h_keys = ['H1','H2','H3','H4','H5','H6']

    for idx, row in df.iterrows():
        model      = row['model']
        technique  = row['technique']
        video      = row['video']
        crime_type = row['crime_type']
        gt         = str(row['ground_truth'])
        output     = str(row['model_output'])
        for judge in judges_for(model):
            ck_key = f'{idx}_{judge}'
            base_row = {
                'row_idx':    idx,
                'model':      model,
                'technique':  technique,
                'video':      video,
                'crime_type': crime_type,
                'judge':      judge,
            }
            if ck_key in ckpt:
                labels = ckpt[ck_key]
                cached_rows.append({
                    **base_row,
                    **{h: labels.get(h, -1) for h in h_keys},
                    'reasoning': labels.get('reasoning', ''),
                })
            else:
                tasks.append((idx, base_row, judge, gt, output, ck_key))

    n_judges = len(JUDGE_FILTER) if JUDGE_FILTER else 2
    print(f'\nPanel: {len(df):,} triplets × {n_judges} judges = {len(tasks)+len(cached_rows):,} calls')
    print(f'Already cached: {len(cached_rows):,}')
    print(f'Remaining:      {len(tasks):,}')
    print(f'Workers:        {max_workers}')

    rows_out.extend(cached_rows)
    if not tasks:
        print('Nothing to do.')
        return pd.DataFrame(rows_out)

    pbar = tqdm(total=len(tasks), desc='Judge calls', unit='call', dynamic_ncols=True)

    def _process_task(task):
        idx, base_row, judge, gt, output, ck_key = task
        try:
            labels = JUDGE_CALL_MAP[judge](gt, output)
        except Exception as e:
            tqdm.write(f'  ERROR [{judge}] row {idx}: {str(e)[:200]}')
            labels = {h: -1 for h in h_keys}
            labels['reasoning'] = f'ERROR: {str(e)[:300]}'
        row_out = {
            **base_row,
            **{h: labels.get(h, -1) for h in h_keys},
            'reasoning': labels.get('reasoning', ''),
        }
        with ckpt_lock:
            ckpt[ck_key] = labels
            if len(ckpt) % 25 == 0:
                save_checkpoint(ckpt)
        with rows_lock:
            rows_out.append(row_out)
        return row_out

    with ThreadPoolExecutor(max_workers=max_workers) as ex:
        futures = [ex.submit(_process_task, t) for t in tasks]
        for _ in as_completed(futures):
            pbar.update(1)
    pbar.close()

    with ckpt_lock:
        save_checkpoint(ckpt)
    return pd.DataFrame(rows_out)

# EXECUTE PANEL
if os.path.exists(PANEL_RAW_PATH):
    df_panel_raw = pd.read_csv(PANEL_RAW_PATH)
    print(f'Loaded existing raw labels: {len(df_panel_raw):,} rows')
else:
    df_panel_raw = run_judge_panel(df_triplets, max_workers=8)
    df_panel_raw.to_csv(PANEL_RAW_PATH, index=False)
    print(f'\nSaved raw labels: {len(df_panel_raw):,} rows')

df_panel_raw.to_csv(JUDGE_LABELS_PATH, index=False)
````

## 10. Inter-judge agreement

````python
def cohens_kappa(y1, y2):
    valid = [(a, b) for a, b in zip(y1, y2) if a in (0, 1) and b in (0, 1)]
    if len(valid) < 2:
        return float('nan')
    a = [v[0] for v in valid]
    b = [v[1] for v in valid]
    n = len(a)
    p_o = sum(x == y for x, y in zip(a, b)) / n
    p_a = (sum(a) / n) * (sum(b) / n) + ((n - sum(a)) / n) * ((n - sum(b)) / n)
    return (p_o - p_a) / (1 - p_a) if p_a < 1 else 1.0

def compute_inter_judge_agreement(df_raw):
    h_cols = ['H1','H2','H3','H4','H5','H6']
    judges = df_raw['judge'].unique().tolist()
    pairs = [(judges[i], judges[j])
             for i in range(len(judges)) for j in range(i+1, len(judges))]
    results = {}
    for j1, j2 in pairs:
        pair_key = f'{j1}_vs_{j2}'
        results[pair_key] = {}
        df1 = df_raw[df_raw['judge'] == j1].set_index('row_idx')
        df2 = df_raw[df_raw['judge'] == j2].set_index('row_idx')
        common_idx = df1.index.intersection(df2.index)
        for h in h_cols:
            y1 = df1.loc[common_idx, h].tolist()
            y2 = df2.loc[common_idx, h].tolist()
            results[pair_key][h] = round(cohens_kappa(y1, y2), 4)
        y1_all = [v for h in h_cols for v in df1.loc[common_idx, h].tolist()]
        y2_all = [v for h in h_cols for v in df2.loc[common_idx, h].tolist()]
        results[pair_key]['overall'] = round(cohens_kappa(y1_all, y2_all), 4)
    return results

print('Computing inter-judge agreement...')
agreement = compute_inter_judge_agreement(df_panel_raw)
with open(JUDGE_AGREEMENT_PATH, 'w') as f:
    json.dump(agreement, f, indent=2)

print('\nInter-judge agreement (Cohen\'s Kappa):')
for pair, kappas in agreement.items():
    print(f'  {pair}:')
    for h, k in kappas.items():
        print(f'    {h}: {k:.4f}')
````

## 11. Aggregate panel votes

````python
def majority_vote(vals):
    valid = [v for v in vals if v in (0, 1)]
    if not valid:
        return -1
    return 1 if sum(valid) >= len(valid) / 2 else 0

def aggregate_panel(df_raw, df_triplets):
    h_cols = ['H1','H2','H3','H4','H5','H6']
    agg = (df_raw
           .groupby(['row_idx','model','technique','video','crime_type'])[h_cols]
           .agg(majority_vote)
           .reset_index())
    inv_map = {v: k for k, v in HTYPE_MAP.items()}
    for h in h_cols:
        agg[inv_map[h]] = agg[h]
    df_t = df_triplets.reset_index().rename(columns={'index': 'row_idx'})
    agg = agg.merge(
        df_t[['row_idx','ground_truth','model_output','model_output_full_len']],
        on='row_idx', how='left'
    )
    agg['hallucination_count'] = agg[h_cols].apply(
        lambda r: sum(v for v in r if v == 1), axis=1
    )
    agg['any_hallucination'] = (agg['hallucination_count'] > 0).astype(int)
    return agg

print('Aggregating votes...')
df_full = aggregate_panel(df_panel_raw, df_triplets)
df_full.to_csv(FULL_LABELED_PATH, index=False)
print(f'Saved: {FULL_LABELED_PATH}  ({len(df_full):,} rows)')
````

# H1 / H5 Disambiguation Pilot

The original prompt's H1 and H5 definitions were ambiguous, producing inter-judge κ
of 0.09–0.34 on those two dimensions vs 0.48–0.82 on H2/H3/H4/H6. Pilot below
disambiguates the definitions and re-judges a small sample.

**Workflow:**
1. Cell 26 — define `JUDGE_SYSTEM_V2` (disambiguated H1/H5 wording, plus examples).
2. Cell 28 — pilot config (subfolder, PILOT_N, etc).
3. Cell 30 — pilot judge functions (uses the new system prompt, reuses SDK clients).
4. Cell 32 — run the pilot panel on the first PILOT_N triplets.
5. Cell 34 — compute pilot κ. If κ for H1/H5 is now ≥ 0.5, the prompt fix worked;
   bump PILOT_N to None for full run. If still < 0.5, the disagreement is not prompt-driven.

All pilot outputs are written to `OUTPUT_DIR / 'h1_h5_disambiguation' /`. The original
`panel_raw_judge_labels_full.csv` and `panel_checkpoint_full.json` are not touched.


````python
JUDGE_SYSTEM_V2 = """You are an expert forensic video analysis evaluator.
Your task is to detect hallucinations in an AI model's analysis of a crime video.
You will be given:
  1. GROUND TRUTH: the human-annotated description of what actually happened.
  2. MODEL OUTPUT: the AI system's analysis of the same video.

Evaluate whether the MODEL OUTPUT contains any of the following hallucination types.
For each type respond ONLY with 0 (not present) or 1 (present).

Hallucination types:

H1 SCENE_FABRICATION - the model adds a SETTING detail that contradicts the ground
   truth, OR names a specific location, structure, or environmental feature that
   ground truth does not mention.
   - Flag (H1=1) example: GT says "a store"; model says "a 7-Eleven convenience store
     with fluorescent lighting." (Adds named features.)
   - Do NOT flag (H1=0) example: GT says "a store"; model says "a small store" or
     "an indoor store." (Generic descriptors are not fabrication.)

H2 CRIME_MISCLASSIFICATION - identifies the wrong crime type relative to ground truth
   (e.g. says Robbery when ground truth is Assault).

H3 CRIME_MISSED - fails to mention the primary crime that is clearly described in
   ground truth.

H4 SEVERITY_MINIMIZATION - describes the crime as substantially less serious than
   ground truth indicates.

H5 ENTITY_FABRICATION - the model introduces a SPECIFIC person, vehicle, or object
   that ground truth does not mention. Specific = identifiable by an attribute,
   role, or name.
   - Flag (H5=1) example: GT mentions "a man"; model says "a man in a red hoodie
     carrying a backpack." (Adds identifying attributes / new objects.)
   - Flag (H5=1) example: GT mentions "a person"; model says "a person and a
     bystander." (Adds a new entity.)
   - Do NOT flag (H5=0) example: GT says "a man"; model says "an adult male" or
     "the suspect." (Rephrasing of the same entity.)

H6 PHANTOM_ACTORS - adds perpetrators or victims not mentioned in ground truth
   (a stricter version of H5 specifically for people in active roles).

Respond ONLY with a JSON object in this exact format, no extra text:
{
  "H1": 0,
  "H2": 0,
  "H3": 0,
  "H4": 0,
  "H5": 0,
  "H6": 0,
  "reasoning": "brief one-sentence explanation"
}"""

print('JUDGE_SYSTEM_V2 (disambiguated H1/H5) defined.')
````

````python
# Pilot config — separate subfolder so v4 outputs remain untouched.
PILOT_DIR = os.path.join(OUTPUT_DIR, 'h1_h5_disambiguation')
os.makedirs(PILOT_DIR, exist_ok=True)

PILOT_CHECKPOINT = os.path.join(PILOT_DIR, 'pilot_checkpoint.json')
PILOT_RAW_PATH   = os.path.join(PILOT_DIR, 'pilot_raw_judge_labels.csv')
PILOT_KAPPA_PATH = os.path.join(PILOT_DIR, 'pilot_inter_judge_agreement.json')

# Number of triplets to run. Start with 200; if pilot kappas improve, set to None
# to run on the full 9,680 triplets.
PILOT_N = 200

print(f'Pilot output dir: {PILOT_DIR}')
print(f'PILOT_N         : {PILOT_N}')
print(f'Pilot checkpoint: {PILOT_CHECKPOINT}')
print(f'Pilot raw CSV   : {PILOT_RAW_PATH}')
print(f'Pilot kappa JSON: {PILOT_KAPPA_PATH}')
````

````python
# Pilot judge functions — same SDK clients & token budgets as v4, but use
# JUDGE_SYSTEM_V2 (disambiguated). Reuse the patched _extract_text helper, retry,
# and parse_judge_response from cells 10/12.

def make_judge_prompt_v2(ground_truth: str, model_output: str) -> str:
    gt_trunc  = ground_truth[:1500]
    out_trunc = model_output[:3000]
    return (
        f"GROUND TRUTH:\n{gt_trunc}\n\n"
        f"MODEL OUTPUT:\n{out_trunc}\n\n"
        "Now evaluate for hallucinations and return the JSON."
    )

def call_claude_judge_v2(ground_truth: str, model_output: str) -> dict:
    prompt = make_judge_prompt_v2(ground_truth, model_output)
    def _call():
        resp = _anthropic_client.messages.create(
            model='claude-opus-4-7',
            max_tokens=8192,
            system=JUDGE_SYSTEM_V2,
            messages=[{'role': 'user', 'content': prompt}],
        )
        return _extract_text_from_anthropic_response(resp)
    raw = _retry_call(_call)
    return parse_judge_response(raw)

def call_gpt_judge_v2(ground_truth: str, model_output: str) -> dict:
    prompt = make_judge_prompt_v2(ground_truth, model_output)
    def _call():
        resp = _openai_client.chat.completions.create(
            model='gpt-5.5',
            messages=[
                {'role': 'system', 'content': JUDGE_SYSTEM_V2},
                {'role': 'user',   'content': prompt},
            ],
            max_completion_tokens=8192,
        )
        return resp.choices[0].message.content
    raw = _retry_call(_call)
    return parse_judge_response(raw)

def call_gemini_judge_v2(ground_truth: str, model_output: str) -> dict:
    prompt = make_judge_prompt_v2(ground_truth, model_output)
    full_prompt = JUDGE_SYSTEM_V2 + '\n\n' + prompt

    _safety = [
        google_genai_types.SafetySetting(category=c, threshold='BLOCK_NONE')
        for c in (
            'HARM_CATEGORY_HARASSMENT',
            'HARM_CATEGORY_HATE_SPEECH',
            'HARM_CATEGORY_SEXUALLY_EXPLICIT',
            'HARM_CATEGORY_DANGEROUS_CONTENT',
        )
    ]
    _gemini_config = google_genai_types.GenerateContentConfig(
        temperature=0.1,
        max_output_tokens=8192,
        top_p=0.8,
        top_k=10,
        safety_settings=_safety,
    )

    def _call():
        try:
            resp = _genai_client.models.generate_content(
                model='gemini-3.1-pro-preview',
                contents=full_prompt,
                config=_gemini_config,
            )
        except Exception as e:
            err = str(e)
            if '429' in err or 'RESOURCE_EXHAUSTED' in err or 'quota' in err.lower():
                tqdm.write('    [Gemini 429] sleeping 60s before retry...')
                time.sleep(60)
            raise

        text = resp.text
        if text is None:
            finish = None
            block_reason = None
            try:
                if resp.candidates:
                    finish = getattr(resp.candidates[0], 'finish_reason', None)
                if getattr(resp, 'prompt_feedback', None):
                    block_reason = getattr(resp.prompt_feedback, 'block_reason', None)
            except Exception:
                pass
            raise ValueError(
                f'Gemini returned no text (finish_reason={finish}, block_reason={block_reason})'
            )
        return text

    raw = _retry_call(_call)
    return parse_judge_response(raw)


JUDGE_CALL_MAP_V2 = {
    'Claude': call_claude_judge_v2,
    'GPT':    call_gpt_judge_v2,
    'Gemini': call_gemini_judge_v2,
}

print('Pilot judge functions ready (using JUDGE_SYSTEM_V2).')
````

````python
# Pilot run — uses its own checkpoint, its own raw CSV, JUDGE_CALL_MAP_V2.
# Reuses the row layout and judges_for() from cell 2.

def load_pilot_checkpoint():
    if os.path.exists(PILOT_CHECKPOINT):
        with open(PILOT_CHECKPOINT) as f:
            return json.load(f)
    return {}

def save_pilot_checkpoint(ckpt):
    with open(PILOT_CHECKPOINT, 'w') as f:
        json.dump(ckpt, f)

def run_pilot_panel(df, judge_call_map, max_workers=8):
    from concurrent.futures import ThreadPoolExecutor, as_completed
    import threading

    ckpt = load_pilot_checkpoint()
    ckpt_lock = threading.Lock()
    rows_lock = threading.Lock()
    rows_out = []
    tasks = []
    cached_rows = []
    h_keys = ['H1','H2','H3','H4','H5','H6']

    for idx, row in df.iterrows():
        model      = row['model']
        technique  = row['technique']
        video      = row['video']
        crime_type = row['crime_type']
        gt         = str(row['ground_truth'])
        output     = str(row['model_output'])
        for judge in judges_for(model):
            ck_key = f'{idx}_{judge}'
            base_row = {
                'row_idx':    idx,
                'model':      model,
                'technique':  technique,
                'video':      video,
                'crime_type': crime_type,
                'judge':      judge,
            }
            if ck_key in ckpt:
                labels = ckpt[ck_key]
                cached_rows.append({
                    **base_row,
                    **{h: labels.get(h, -1) for h in h_keys},
                    'reasoning': labels.get('reasoning', ''),
                })
            else:
                tasks.append((idx, base_row, judge, gt, output, ck_key))

    print(f'\nPilot panel: {len(df):,} triplets')
    print(f'Already cached: {len(cached_rows):,}')
    print(f'Remaining:      {len(tasks):,}')
    print(f'Workers:        {max_workers}')

    rows_out.extend(cached_rows)
    if not tasks:
        print('Nothing to do.')
        return pd.DataFrame(rows_out)

    pbar = tqdm(total=len(tasks), desc='Pilot judge calls', unit='call', dynamic_ncols=True)

    def _process_task(task):
        idx, base_row, judge, gt, output, ck_key = task
        try:
            labels = judge_call_map[judge](gt, output)
        except Exception as e:
            tqdm.write(f'  ERROR [{judge}] row {idx}: {str(e)[:200]}')
            labels = {h: -1 for h in h_keys}
            labels['reasoning'] = f'ERROR: {str(e)[:300]}'
        row_out = {
            **base_row,
            **{h: labels.get(h, -1) for h in h_keys},
            'reasoning': labels.get('reasoning', ''),
        }
        with ckpt_lock:
            ckpt[ck_key] = labels
            if len(ckpt) % 25 == 0:
                save_pilot_checkpoint(ckpt)
        with rows_lock:
            rows_out.append(row_out)
        return row_out

    with ThreadPoolExecutor(max_workers=max_workers) as ex:
        futures = [ex.submit(_process_task, t) for t in tasks]
        for _ in as_completed(futures):
            pbar.update(1)
    pbar.close()

    with ckpt_lock:
        save_pilot_checkpoint(ckpt)
    return pd.DataFrame(rows_out)


# Pick the pilot subset
if PILOT_N is None:
    df_pilot = df_triplets.copy()
    print(f'Running FULL pilot on {len(df_pilot):,} triplets')
else:
    df_pilot = df_triplets.head(PILOT_N).copy()
    print(f'Running PILOT on first {PILOT_N} triplets')

df_pilot_raw = run_pilot_panel(df_pilot, JUDGE_CALL_MAP_V2, max_workers=8)
df_pilot_raw.to_csv(PILOT_RAW_PATH, index=False)
print(f'\nSaved pilot raw labels: {PILOT_RAW_PATH}')
print(f'  Rows: {len(df_pilot_raw):,}')
````

````python
# Compute pilot kappas using the existing cohens_kappa from cell 20.
# Compare to the v4 main-panel kappas to see if disambiguation helped.

df_pilot = pd.read_csv(PILOT_RAW_PATH)
h_cols = ['H1','H2','H3','H4','H5','H6']

# Filter out -1 rows and pivot wide so we have one row per row_idx with three judge columns
valid = df_pilot[df_pilot[h_cols].apply(lambda r: all(v >= 0 for v in r), axis=1)]
print(f'Valid pilot rows: {len(valid):,} of {len(df_pilot):,}')

# For each pair of judges, compute kappa per dimension and overall
pilot_kappas = {}
for j1, j2 in [('GPT','Gemini'), ('GPT','Claude'), ('Gemini','Claude')]:
    pair_key = f'{j1}_vs_{j2}'
    pivot = valid.pivot_table(index='row_idx', columns='judge', values=h_cols)
    pivot = pivot.dropna()
    if j1 not in pivot.columns.get_level_values(1) or j2 not in pivot.columns.get_level_values(1):
        print(f'  {pair_key}: no overlapping rows yet')
        continue
    per_dim = {}
    for h in h_cols:
        try:
            y1 = pivot[(h, j1)].astype(int).values
            y2 = pivot[(h, j2)].astype(int).values
            per_dim[h] = round(cohens_kappa(y1, y2), 4)
        except KeyError:
            per_dim[h] = None
    # Overall kappa: flatten across all H dims
    try:
        all_y1 = []
        all_y2 = []
        for h in h_cols:
            all_y1.extend(pivot[(h, j1)].astype(int).values)
            all_y2.extend(pivot[(h, j2)].astype(int).values)
        per_dim['overall'] = round(cohens_kappa(all_y1, all_y2), 4)
    except KeyError:
        per_dim['overall'] = None
    pilot_kappas[pair_key] = per_dim

# Save
with open(PILOT_KAPPA_PATH, 'w') as f:
    json.dump(pilot_kappas, f, indent=2)

print(f'\nSaved pilot kappas: {PILOT_KAPPA_PATH}\n')
print('Pilot inter-judge kappa (disambiguated H1/H5):')
print('=' * 75)
hdr = f'{"Pair":22s}  ' + '  '.join(f'{h:>7s}' for h in h_cols + ['overall'])
print(hdr)
print('-' * 75)
for pair, vals in pilot_kappas.items():
    line = f'{pair:22s}  ' + '  '.join(
        f'{vals.get(h, float("nan")):>7.3f}' if vals.get(h) is not None else f'{"--":>7s}'
        for h in h_cols + ['overall']
    )
    print(line)

print()
print('=' * 75)
print('DECISION GATE')
print('=' * 75)
# Look at H1/H5 specifically across all pairs
h15_pairs = []
for pair, vals in pilot_kappas.items():
    if 'Claude' in pair:
        h15_pairs.append((pair, 'H1', vals.get('H1')))
        h15_pairs.append((pair, 'H5', vals.get('H5')))
for pair, dim, k in h15_pairs:
    if k is None:
        continue
    tag = 'GOOD' if k >= 0.5 else ('MARGINAL' if k >= 0.3 else 'STILL LOW')
    print(f'  {pair:22s} {dim}: kappa={k:.3f}  -> {tag}')
print()
print('If H1/H5 kappas vs Claude are now >= 0.5: prompt fix worked. Set PILOT_N = None')
print('and re-run cell 32 for the full re-judging.')
print('If still low: the disagreement is conceptual, not prompt-driven. Report as a finding.')
````

# H1 / H5 Pilot — Iteration 2 (middle-ground prompt + stratified sample)

The v5 pilot showed:
- **Original prompt**: GPT/Gemini agree well on H1/H5 (κ ~0.75) but Claude is an outlier (κ ~0.1–0.3).
- **Strict prompt (v5)**: GPT/Gemini both flag ~100% of rows on H1/H5 (no variance, κ undefined). Also missed Claude (sample was Claude-only).

This iteration:
1. **Middle-ground prompt** with paired flag/no-flag examples drawn from concrete UCF-Crime cases.
2. **Stratified sample** (~67 rows × 3 models = ~201) so all three judges contribute.

Outputs go to `OUTPUT_DIR/h1_h5_v3/` — separate from v5's `h1_h5_disambiguation/`.


````python
JUDGE_SYSTEM_V3 = """You are an expert forensic video analysis evaluator.
Your task is to detect hallucinations in an AI model's analysis of a crime video.
You will be given:
  1. GROUND TRUTH: the human-annotated description of what actually happened.
  2. MODEL OUTPUT: the AI system's analysis of the same video.

Evaluate whether the MODEL OUTPUT contains the following hallucination types.
For each type respond ONLY with 0 (not present) or 1 (present).

Important calibration: ground truth descriptions are intentionally terse. The model
output is expected to add visual detail. Only flag fabrication when the model adds
*meaningfully new claims* — not when it is simply rephrasing or making generic
descriptive statements consistent with the ground truth.

H1 SCENE_FABRICATION - flag (=1) ONLY if the model output asserts a setting detail
   that EITHER (a) contradicts ground truth, OR (b) names a specific identifiable
   place, brand, or object that adds factual content beyond what ground truth
   establishes (and could not be inferred without seeing more than was said).
   - FLAG: GT says "store"; output says "7-Eleven on Main Street with neon sign."
     (Adds named identifiable place + specific brand.)
   - FLAG: GT says "outdoor area"; output says "fenced parking lot with three cars."
     (Asserts specific count, contradictable detail.)
   - DO NOT FLAG: GT says "store"; output says "small store" or "indoor store" or
     "convenience store." (Generic descriptors / category-level specification.)
   - DO NOT FLAG: GT says "street"; output says "city street" or "narrow street."
     (Generic descriptors.)

H2 CRIME_MISCLASSIFICATION - identifies the wrong crime type relative to ground
   truth (e.g. says Robbery when ground truth is Assault).

H3 CRIME_MISSED - fails to mention the primary crime that is clearly described in
   ground truth.

H4 SEVERITY_MINIMIZATION - describes the crime as substantially less serious than
   ground truth indicates.

H5 ENTITY_FABRICATION - flag (=1) ONLY if the model output asserts a SPECIFIC
   identifiable entity (named, attributed, counted, or with distinguishing features)
   that is NOT in ground truth.
   - FLAG: GT mentions "a man"; output says "a man wearing a red hoodie carrying a
     black backpack." (Adds identifying attributes that constitute new factual claims.)
   - FLAG: GT mentions "a person"; output says "a person and a security guard."
     (Adds a new entity with a role.)
   - FLAG: GT mentions "a vehicle"; output says "a white Toyota Camry." (Specific
     make/model/color is a new factual claim.)
   - DO NOT FLAG: GT says "a man"; output says "an adult male" or "the suspect" or
     "a person." (Synonymous rephrasing of the same entity.)
   - DO NOT FLAG: GT says "a vehicle"; output says "a car" or "a passenger vehicle."
     (Generic category descriptors.)

H6 PHANTOM_ACTORS - adds perpetrators or victims (people in active crime roles) not
   mentioned in ground truth. Stricter than H5: must be a person playing an active
   role in the incident, not just any new entity.

Respond ONLY with a JSON object in this exact format, no extra text:
{
  "H1": 0,
  "H2": 0,
  "H3": 0,
  "H4": 0,
  "H5": 0,
  "H6": 0,
  "reasoning": "brief one-sentence explanation"
}"""

print('JUDGE_SYSTEM_V3 (middle-ground H1/H5 with paired examples) defined.')
````

````python
# Build the stratified sample. Sample PILOT_V3_PER_MODEL rows from each generator model.
import numpy as np

if os.path.exists(PILOT_V3_SAMPLE_PATH):
    df_pilot_v3 = pd.read_csv(PILOT_V3_SAMPLE_PATH)
    print(f'Loaded existing stratified sample: {len(df_pilot_v3):,} triplets')
else:
    parts = []
    for model in ['Claude', 'GPT', 'Gemini']:
        sub = df_triplets[df_triplets['model'] == model]
        if len(sub) < PILOT_V3_PER_MODEL:
            print(f'  WARNING: only {len(sub)} {model} rows available, taking all')
            parts.append(sub)
        else:
            parts.append(sub.sample(n=PILOT_V3_PER_MODEL, random_state=RANDOM_SEED))
    df_pilot_v3 = pd.concat(parts).reset_index(drop=False)  # keep original row_idx
    df_pilot_v3 = df_pilot_v3.rename(columns={'index': 'orig_row_idx'})
    df_pilot_v3.to_csv(PILOT_V3_SAMPLE_PATH, index=False)
    print(f'Built stratified sample: {len(df_pilot_v3):,} triplets')

print(f'\nSample composition:')
print(df_pilot_v3.groupby(['model','technique']).size().unstack(fill_value=0))
print(f'\nExpected judge calls (each row gets 2 cross-judges): {len(df_pilot_v3)*2:,}')
print(f'Expected per judge: ~{len(df_pilot_v3)*2//3:,}')
````

````python
# Judge functions v3 — same SDK calls as v4, just with JUDGE_SYSTEM_V3.

def make_judge_prompt_v3(ground_truth: str, model_output: str) -> str:
    gt_trunc  = ground_truth[:1500]
    out_trunc = model_output[:3000]
    return (
        f"GROUND TRUTH:\n{gt_trunc}\n\n"
        f"MODEL OUTPUT:\n{out_trunc}\n\n"
        "Now evaluate for hallucinations and return the JSON."
    )

def call_claude_judge_v3(ground_truth, model_output):
    prompt = make_judge_prompt_v3(ground_truth, model_output)
    def _call():
        resp = _anthropic_client.messages.create(
            model='claude-opus-4-7',
            max_tokens=8192,
            system=JUDGE_SYSTEM_V3,
            messages=[{'role': 'user', 'content': prompt}],
        )
        return _extract_text_from_anthropic_response(resp)
    raw = _retry_call(_call)
    return parse_judge_response(raw)

def call_gpt_judge_v3(ground_truth, model_output):
    prompt = make_judge_prompt_v3(ground_truth, model_output)
    def _call():
        resp = _openai_client.chat.completions.create(
            model='gpt-5.5',
            messages=[
                {'role': 'system', 'content': JUDGE_SYSTEM_V3},
                {'role': 'user',   'content': prompt},
            ],
            max_completion_tokens=8192,
        )
        return resp.choices[0].message.content
    raw = _retry_call(_call)
    return parse_judge_response(raw)

def call_gemini_judge_v3(ground_truth, model_output):
    prompt = make_judge_prompt_v3(ground_truth, model_output)
    full_prompt = JUDGE_SYSTEM_V3 + '\n\n' + prompt
    _safety = [
        google_genai_types.SafetySetting(category=c, threshold='BLOCK_NONE')
        for c in ('HARM_CATEGORY_HARASSMENT','HARM_CATEGORY_HATE_SPEECH',
                  'HARM_CATEGORY_SEXUALLY_EXPLICIT','HARM_CATEGORY_DANGEROUS_CONTENT')
    ]
    _gemini_config = google_genai_types.GenerateContentConfig(
        temperature=0.1, max_output_tokens=8192, top_p=0.8, top_k=10,
        safety_settings=_safety,
    )
    def _call():
        try:
            resp = _genai_client.models.generate_content(
                model='gemini-3.1-pro-preview', contents=full_prompt, config=_gemini_config,
            )
        except Exception as e:
            err = str(e)
            if '429' in err or 'RESOURCE_EXHAUSTED' in err or 'quota' in err.lower():
                tqdm.write('    [Gemini 429] sleeping 60s before retry...')
                time.sleep(60)
            raise
        text = resp.text
        if text is None:
            finish = None; block_reason = None
            try:
                if resp.candidates:
                    finish = getattr(resp.candidates[0], 'finish_reason', None)
                if getattr(resp, 'prompt_feedback', None):
                    block_reason = getattr(resp.prompt_feedback, 'block_reason', None)
            except Exception:
                pass
            raise ValueError(f'Gemini returned no text (finish_reason={finish}, block_reason={block_reason})')
        return text
    raw = _retry_call(_call)
    return parse_judge_response(raw)

JUDGE_CALL_MAP_V3 = {
    'Claude': call_claude_judge_v3,
    'GPT':    call_gpt_judge_v3,
    'Gemini': call_gemini_judge_v3,
}

print('Pilot v3 judge functions ready.')
````

````python
# Run pilot v3 panel
def load_pilot_v3_ckpt():
    if os.path.exists(PILOT_V3_CHECKPOINT):
        with open(PILOT_V3_CHECKPOINT) as f:
            return json.load(f)
    return {}

def save_pilot_v3_ckpt(ckpt):
    with open(PILOT_V3_CHECKPOINT, 'w') as f:
        json.dump(ckpt, f)

def run_pilot_v3(df, max_workers=8):
    from concurrent.futures import ThreadPoolExecutor, as_completed
    import threading
    ckpt = load_pilot_v3_ckpt()
    ckpt_lock = threading.Lock()
    rows_lock = threading.Lock()
    rows_out = []
    tasks = []
    cached_rows = []
    h_keys = ['H1','H2','H3','H4','H5','H6']

    for _, row in df.iterrows():
        # Use orig_row_idx as the stable key (matches df_triplets index)
        idx = row.get('orig_row_idx', row.name)
        model      = row['model']
        technique  = row['technique']
        video      = row['video']
        crime_type = row['crime_type']
        gt         = str(row['ground_truth'])
        output     = str(row['model_output'])
        for judge in judges_for(model):
            ck_key = f'{idx}_{judge}'
            base_row = {
                'row_idx':    idx,
                'model':      model,
                'technique':  technique,
                'video':      video,
                'crime_type': crime_type,
                'judge':      judge,
            }
            if ck_key in ckpt:
                labels = ckpt[ck_key]
                cached_rows.append({**base_row,
                                    **{h: labels.get(h, -1) for h in h_keys},
                                    'reasoning': labels.get('reasoning', '')})
            else:
                tasks.append((idx, base_row, judge, gt, output, ck_key))

    print(f'\nPilot v3: {len(df):,} triplets')
    print(f'Already cached: {len(cached_rows):,}')
    print(f'Remaining:      {len(tasks):,}')
    rows_out.extend(cached_rows)
    if not tasks:
        print('Nothing to do.')
        return pd.DataFrame(rows_out)

    pbar = tqdm(total=len(tasks), desc='Pilot v3 calls', unit='call', dynamic_ncols=True)
    def _process_task(task):
        idx, base_row, judge, gt, output, ck_key = task
        try:
            labels = JUDGE_CALL_MAP_V3[judge](gt, output)
        except Exception as e:
            tqdm.write(f'  ERROR [{judge}] row {idx}: {str(e)[:200]}')
            labels = {h: -1 for h in h_keys}
            labels['reasoning'] = f'ERROR: {str(e)[:300]}'
        row_out = {**base_row,
                   **{h: labels.get(h, -1) for h in h_keys},
                   'reasoning': labels.get('reasoning', '')}
        with ckpt_lock:
            ckpt[ck_key] = labels
            if len(ckpt) % 25 == 0:
                save_pilot_v3_ckpt(ckpt)
        with rows_lock:
            rows_out.append(row_out)
        return row_out

    with ThreadPoolExecutor(max_workers=max_workers) as ex:
        futures = [ex.submit(_process_task, t) for t in tasks]
        for _ in as_completed(futures):
            pbar.update(1)
    pbar.close()
    with ckpt_lock:
        save_pilot_v3_ckpt(ckpt)
    return pd.DataFrame(rows_out)

df_pilot_v3_raw = run_pilot_v3(df_pilot_v3, max_workers=8)
df_pilot_v3_raw.to_csv(PILOT_V3_RAW_PATH, index=False)
print(f'\nSaved: {PILOT_V3_RAW_PATH}')
print(f'  Rows: {len(df_pilot_v3_raw):,}')
````

````python
# Compute pilot v3 kappas + per-judge positive rate diagnostics.
df = pd.read_csv(PILOT_V3_RAW_PATH)
h_cols = ['H1','H2','H3','H4','H5','H6']

valid = df[df[h_cols].apply(lambda r: all(v >= 0 for v in r), axis=1)]
print(f'Valid pilot v3 rows: {len(valid):,} of {len(df):,}')

print('\nPer-judge row counts:')
print(valid['judge'].value_counts())

print('\nPositive rate per judge per dimension:')
rate = valid.groupby('judge')[h_cols].mean().round(3)
print(rate)

# Pairwise kappa: align on common row_idx
pilot_kappas = {}
for j1, j2 in [('GPT','Gemini'), ('GPT','Claude'), ('Gemini','Claude')]:
    pair_key = f'{j1}_vs_{j2}'
    j1_rows = valid[valid['judge'] == j1].set_index('row_idx')
    j2_rows = valid[valid['judge'] == j2].set_index('row_idx')
    common = j1_rows.index.intersection(j2_rows.index)
    if len(common) == 0:
        print(f'\n{pair_key}: no common rows')
        pilot_kappas[pair_key] = None
        continue
    print(f'\n{pair_key}: {len(common)} common rows')
    j1a = j1_rows.loc[common]
    j2a = j2_rows.loc[common]
    per_dim = {}
    for h in h_cols:
        y1 = j1a[h].astype(int).values
        y2 = j2a[h].astype(int).values
        pos1 = y1.mean()
        pos2 = y2.mean()
        agree = (y1 == y2).mean()
        if len(set(y1)) == 1 or len(set(y2)) == 1:
            note = f'(constant: {j1}={pos1:.0%}, {j2}={pos2:.0%}, raw agreement={agree:.1%})'
            per_dim[h] = None
        else:
            try:
                k = round(cohens_kappa(y1, y2), 4)
                per_dim[h] = k
                note = f'kappa={k}, agreement={agree:.1%}'
            except Exception:
                per_dim[h] = None
                note = '(error)'
        print(f'  {h}: {note}')
    # Overall (concatenated)
    all_y1 = []; all_y2 = []
    for h in h_cols:
        all_y1.extend(j1a[h].astype(int).values)
        all_y2.extend(j2a[h].astype(int).values)
    try:
        per_dim['overall'] = round(cohens_kappa(all_y1, all_y2), 4)
    except Exception:
        per_dim['overall'] = None
    pilot_kappas[pair_key] = per_dim

with open(PILOT_V3_KAPPA_PATH, 'w') as f:
    json.dump(pilot_kappas, f, indent=2)
print(f'\nSaved kappas: {PILOT_V3_KAPPA_PATH}')

print('\n' + '=' * 75)
print('DECISION GATE')
print('=' * 75)
# Specifically check H1 and H5 across all 3 pairs
ok_h15 = True
for pair, vals in pilot_kappas.items():
    if vals is None:
        continue
    for h in ('H1', 'H5'):
        k = vals.get(h)
        tag = 'GOOD' if (k is not None and k >= 0.5) else \
              ('CONSTANT' if k is None else \
               ('MARGINAL' if k >= 0.3 else 'STILL LOW'))
        print(f'  {pair:22s} {h}: {k}  -> {tag}')
        if k is None or k < 0.5:
            ok_h15 = False

print()
if ok_h15:
    print('All H1/H5 pairs >= 0.5. Prompt v3 works. Set sample size to None and re-run')
    print('cell that builds df_pilot_v3 (delete sample CSV first) for full panel.')
else:
    print('Some H1/H5 pairs still low or constant. Either:')
    print('  - Prompt needs further iteration, or')
    print('  - Disagreement is conceptual, not prompt-driven.')
    print('Recommend: stop iterating on the prompt; report v4 H1/H5 disagreement as a finding.')
````
